Break and Continue statements 

```python
for i in range(100,200):
    fact=True
    for j in range(2,int(i**0.5)+1):
        if i%j==0:
            fact=False
            break
    if fact:
        print("First prime number greater than 100 is:",i)
        break
```

```python
s=0
for i in range(1,101):
    if i%3==0 and i%5==0:
        continue
    s+=i
print("Sum:",s)
```

```python
n=int(input())
l=[]
x=1
while True:
    if x>n:
        break
    if n%x==0:
        l.append(x)
    x+=1
print(*l)
```

Pass statements

```python
l=list(map(int,input().split()))
for i in l:
    if i%2!=0:
        pass
    else:
        print(i)
```

```python
l=list(map(int,input().split()))
for i in l:
    fact=0
    for j in range(1,i+1):
        if i%j==0:
            fact+=1
    if fact==2:
        print(i)
    else:
        pass
```
