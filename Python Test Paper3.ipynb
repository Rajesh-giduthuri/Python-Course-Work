```python
#1.Sum of Natural Numbers
n=int(input("Enter N: "))
print(f"Sum: {n*(n+1)//2}")
```

```python
#1.Sum of Natural Numbers(using Recursion)-
def Natural_sum(n):
    if n==0:
        return 0
    s=n+Natural_sum(n-1)
    return s

n=int(input("Enter N: "))
print(f"Sum: {Natural_sum(n)}")
```

```python
#2.Count Digits using Loops
n=int(input("Enter Number: "))
count=0
while n>0:
    count+=1
    n=n//10
print(f"Digit Count: {count}")
```

```python
#2.Count Digits using Loops(using Recursion)
def Digit_count(n):
    if n==0:
        return 0
    return 1+Digit_count(n//10)

n=int(input("Enter Number: "))
print(f"Digit Count: {Digit_count(n)}")
```
