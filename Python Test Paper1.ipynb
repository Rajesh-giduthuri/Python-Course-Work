```python
#1) Password Strength Checker
s=input()
digit=0
upper=0
flag=False
if len(s)>=8:
    for i in s:
        if i.isdigit():
            digit+=1
        if i.isupper():
            upper+=1
    if digit>0 and upper>0:
        flag=True
print("Password is Strong" if flag==True else "Password is Weak")
```

```python
#1) Password Strength Checker
#any() uses a generator, so it doesn't create a new list or copy of the string.
#Optimized
s = input()
flag = len(s) >= 8 and any(c.isdigit() for c in s) and any(c.isupper() for c in s)
print("Password is Strong" if flag else "Password is Weak")
```

```python
#2) Shopping Cart System
def calculate_bill(l):
    total=sum(l)
    bill=total
    if total>3000:
        discount="10%"
        bill-=(0.1*total)
    else:
        discount="no discount"
    return total,discount,bill
    
print("Enter number of products: ",end="")
n=int(input())
l=[]
for i in range(1,n+1):
    price = int(input(f"Enter price {i}: "))
    l.append(price)

tup=calculate_bill(l)
print("Total Amount: ",tup[0])
print("Discount: ",tup[1])
print(f"Final Bill to pay: {tup[2]:.2f}")

```

```python
#3) Bank Account System 
class BankAccount:
    def __init__(self):
        self.bal=0
    def deposit(self,dep):
        #self.dep=dep
        self.bal+=dep
        print("Deposit successfull!")
    def withdraw(self,draw):
        #self.draw=draw
        if draw>self.bal:
            print("Can't withdraw amount!")
            print(f"Available balance: ₹{self.bal}")

        elif self.bal-draw<1000:
            print("Can't withdraw amount!")
            print("Minimum balance of ₹1000 must be maintained.")
            print(f"Available balance: ₹{self.bal}")

        else:
            self.bal-=draw
            print("Withdraw successful!")
            print(f"Remaining balance: ₹{self.bal}")
    def check_balance(self):
        print(f"Available balance: ₹{self.bal}")

obj=BankAccount()
print("Enter Initial Amount: ₹",end="")
initial_amount=int(input())
obj.deposit(initial_amount)

print("1. Deposit")
print("2. Withdraw")
print("3. Balance")
print("Enter Operation to perform: ",end="")
operation=int(input())
if operation==1:
    print("Enter Deposit Amount: ₹",end="")
    amount=int(input())
    obj.deposit(amount)
elif operation==2:
    print("Enter Withdrawal Amount: ₹",end="")
    amount=int(input())
    obj.withdraw(amount)
elif operation==3:
    obj.check_balance()
else:
    print("Invalid Operation!")
```

```python
#4) Restaurant Order Analyzer
def Restaurant_Order_Analyzer(l,n):
    if n==0:
        return 0,0,0
    total,max_order,count=Restaurant_Order_Analyzer(l,n-1)
    total+=l[n-1]
    if l[n-1]>max_order:
        max_order=l[n-1]
    if l[n-1]>500:
        count+=1
    return total,max_order,count
    
print("Enter number of orders: ",end="")
n=int(input())
print("Enter order amounts: ",end="")
l=list(map(int,input().split()))

tup=Restaurant_Order_Analyzer(l,n)
print("Total Revenue: ",tup[0])
print("Highest Order: ",tup[1])
print("Orders above 500: ",tup[2])
```
