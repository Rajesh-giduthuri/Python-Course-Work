```python
#1. Print a Hollow Diamond Pattern
n=int(input())
for i in range(1,n+1):
    print(" "*(n-i),end=" ")
    if i==1:
        print("*")
    else:
        print("*"+" "*(2*i-3)+"*")
for i in range(n-1,0,-1):
    print(" "*(n-i),end=" ")
    if i==1:
        print("*")
    else:
        print("*"+" "*(2*i-3)+"*")
        
```

```python
#2. Generate the Power Set of a Given Set

from itertools import combinations

arr = [1, 2, 3]


for i in range(len(arr) + 1):
    for j in combinations(arr, i):
        if len(j)==0:
            print({})
        else:
            print(set(j))
```
