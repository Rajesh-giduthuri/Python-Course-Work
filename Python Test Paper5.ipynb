```python
#1.Uppercase Counter
s=input("Enter string: ")
cou=0
for c in s:
    if c.isupper():
        cou+=1
print("Uppercase Letters: ",cou)
      
```

```python
#1.Uppercase Counter useing ASCII values
s=input("Enter string: ")
cou=0
for c in s:
    if 65<=ord(c)<=90:
        cou+=1
print("Uppercase Letters: ",cou)
```

```python
#2.Average Temperature
l=list(map(int,input("Enter temperatures: ").split()))
print(f"Average Temperature: {(sum(l)/len(l)):.2f}")
```

```python
#3.Factorial using recursion
def fact(n):
    if n==0:
        return 1
    return n*fact(n-1)

n=int(input("Enter number: "))
print("Factorial:",fact(n))
```

```python
#4. Movie Ticket System Implementation using Class
class Movie_Ticket_System:
    def booking(self,p,t):
        amount=t*p
        if t>5:
            amount*=0.90
        return int(amount)

p=int(input("Enter ticket price: "))
t=int(input("Enter number of tickets: "))
obj=Movie_Ticket_System()
print("Final Amount:",obj.booking(p,t))
```
