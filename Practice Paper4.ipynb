```python
#1. Generate Fibonacci Numbers
def fib(n):
    a=0
    b=1
    l=[]
    if n==1:
        l.append(a)
    elif n==2:
        l.append(a)
        l.append(b)
    else:
        l.append(a)
        l.append(b)
        for i in range(3,n+1):
            c=a+b
            l.append(c)
            a=b
            b=c
    return l
n=int(input())
print(*(fib(n)))
```

```python
#1. Generate Fibonacci Numbers Using Recursion
def fib(n):
    if n<=1:
        return n
    return fib(n-1)+fib(n-2)
n=int(input())
for i in range(n):
    print(fib(i),end=" ")
        
```

```python
#2. Check Whether a String is a Palindrome
def pal(s):
    if s==s[::-1]:
        return "Palindrome"
    else:
        return "Not Palindrome"
s=input()
print(pal(s))        
```

```python
#2. Check Whether a String is a Palindrome Using Recursion
def pal(s):
    if len(s)<=1:
        return "Palindrome"
    if s[0]!=s[-1]:
        return "Not Palindrome"
    else:
        return pal(s[1:-1])
s=input()
print(pal(s))
    
```

```python
#3. Generate an Infinite Fibonacci Sequence Using a Generator
def fibonacci():
    a,b=0,1
    while True:
        yield a  
        a,b=b,a+b   

n=int(input())

gen=fibonacci()
for i in range(n):
    print(next(gen),end=" ")
```
