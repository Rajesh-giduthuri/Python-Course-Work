```python
#1. Parking Fee
h=int(input("Enter Parking Hours: "))
fee=20*h  
if h>10:
    print("Parking exceeds 10 hours, an extra ₹100 maintenance fee is added")
    fee+=100
print(f"Total Parking Fee: ₹{fee}")
```

```python
#2. Salary Bonus
def calculate_salary(s):
    if s>=50000:
        return s*1.10
    elif s>=30000:
        return s*1.05
    else:
        return s

s=int(input("Enter salary: "))
print(f"Final Salary: {int(calculate_salary(s))}")       
```

```python
#3. Sum of Digits
def even_nums_addition(n):
    if n<=0:
        return 0
    if n%2==0:
        return n+even_nums_addition(n-1)
    return even_nums_addition(n-1)

n=int(input("Enter number: "))
print(f"Sum of even numbers: {even_nums_addition(n)}")
```

```python
#4.Product Billing System using Class and object
class BillingSystem:
    def calculate_final_bill(self,price,quantity):
        total=price*quantity
        if total>5000:
            total*=0.85
        return int(total)
        
prod=input("Enter product name: ")
p=int(input("Enter price: "))
q=int(input("Enter quantity: "))
c=BillingSystem()
print(f"Product: {prod}")
print(f"Final Bill: {c.calculate_final_bill(p,q)}")
```
