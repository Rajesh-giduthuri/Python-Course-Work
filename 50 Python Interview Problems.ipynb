```python
#41. Longest consecutive sequence
n=int(input())
l=list(map(int,input().split()))
l.sort()
count=1
long=1
for i in range(1,n):
    if l[i]==l[i-1]+1:
        count+=1
    elif l[i]!=l[i-1]:
        count=1
    if count>long:
        long=count
print(long)
```

```python
#42. Majority element
n=int(input())
l=list(map(int,input().split()))
for i in l:
    count=0
    for j in l:
        if i==j:
            count+=1
    if count>n//2:
        print(i)
        break
```

```python
#42. Majority element
n=int(input())
l=list(map(int,input().split()))
for i in l:
    if l.count(i)>n//2:
        print(i)
        break
```

```python
#43. Leaders in an array
n=int(input())
l=list(map(int,input().split()))
for i in range(n):
    lead=True
    for j in range(i+1,n):
        if l[i]<l[j]:
            lead=False
    if lead:
        print(l[i],end=" ")
```

```python
#43. Leaders in an array
n=int(input())
l=list(map(int,input().split()))
maxi=l[n-1]
res=[maxi]
i=l[n-2]
while i>=0:
    if l[i]>maxi:
        maxi=l[i]
        res.append(maxi)
    i-=1
print(*res[::-1])
```

```python
#44. Equilibrium index
n=int(input())
l=list(map(int,input().split()))
for i in range(n):
    left=0
    right=0
    for j in range(i):
        left+=l[j]
    for k in range(i+1,n):
        right+=l[k]
    if left==right:
        print(i)
        break
else:
    print(-1)
```

```python
#44. Equilibrium index
n=int(input())
l=list(map(int,input().split()))
total=sum(l)
left=0
for i in range(n):
    right=total-left-l[i]
    if left==right:
        print(i)
        break
    left+=l[i]
else:
    print(-1)
```

```python
#45. Maximum difference
n=int(input())
l=list(map(int,input().split()))
max_diff=l[1]-l[0]
for i in range(0,n):
    for j in range(i+1,n):
        diff=l[j]-l[i]
        if diff>max_diff:
            max_diff=diff
print(max_diff)
```

```python
#45. Maximum difference
n=int(input())
l=list(map(int,input().split()))
mini=l[0]
max_diff=l[1]-l[0]
for i in range(n):
    diff=l[i]-mini
    if diff>max_diff:
        max_diff=diff
    if l[i]<mini:
        mini=l[i]
print(max_diff)
```

```python
#46. Longest substring without repeating characters
s = input()
longest = 0

for i in range(len(s)):
    sub = ""
    for j in range(i, len(s)):
        if s[j] not in sub:
            sub += s[j]
        else:
            break
    longest = max(longest, len(sub))

print(longest)
```

```python
#49. Maximum perimeter triangle
#Once sorted in descending order, checking consecutive triples is sufficient. 
#The first valid triple gives the maximum perimeter.
n = int(input())
l = list(map(int, input().split()))
l.sort(reverse=True)

for i in range(n-2):
    a = l[i]
    b = l[i+1]
    c = l[i+2]
    if b + c > a:
        print(c, b, a)
        break
else:
    print(-1)
```

```python
#50. Minimum swaps to group all 1s together
n = int(input())
l = list(map(int, input().split()))
ones = l.count(1)
zeros = 0

for i in range(ones):
    if l[i] == 0:
        zeros += 1

res = zeros
for i in range(ones, n):
    if l[i-ones] == 0:
        zeros -= 1
    if l[i] == 0:
        zeros += 1

    res = min(res, zeros)

print(res)
```
