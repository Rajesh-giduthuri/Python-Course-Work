```python
#1. Multiplication Table – School Mathematics Tool
n=int(input())
for i in range(1,11):
    print(f"{n} x {i} = {n*i}",end="\n")
```

```python
#2. Reverse a Number – ATM Security Verification
x=int(input())
r=0
while x>0:
    d=x%10
    r=r*10+d
    x=x//10
print(r)    
```

```python
#sum of digits of a number
n=int(input())
s=0
while n>0:
    d=n%10
    s+=d
    n=n//10
print(s)
```

```python
#Prime or not Prime
n=int(input())
cou=0
for i in range(1,n+1):
    if n%i==0:
        cou+=1
if cou==2:
    print("Prime")
else:
    print("Not Prime")
```

```python
#3. Prime Numbers from 1 to 100 – Security Number Checker
a,b=int(input()),int(input())
l=[]
cou=0
for i in range(a,b):
    fact=0
    for j in range(1,i+1):
        if i%j==0:
            fact+=1
    if fact==2:
        l.append(i)
        cou+=1
        
print(l)
print(cou)
        
```

```python
#4. Count Vowels and Consonants – Student Text Analyzer
s=input().replace(" ","").lower()
vowels=0
consonents=0
for i in s:
    if i.isalpha():
        if i in 'aeiou':
            vowels+=1
        else:
            consonents+=1

print(f"Vowels: {vowels}")
print(f"Consonents: {consonents}")
```

```python
#5. Reverse a String Using a Function – Message Processing System
def reverse_string(s):
    new="".join(s[::-1])
    return new

s=input()
print(reverse_string(s))
```

```python
#5. Reverse a String Using a Function – Message Processing System
def reverse_string(s):
    rev=""
    for c in s:
        rev=c+rev
    return rev

s=input()
print(reverse_string(s))
```
