```python
#1. Design an Airline Dynamic Pricing Engine
s=input().lower()
d=int(input())
f=input().lower()=='true'
a=int(input())
bp=5000

if s=="business":
    bp*=1.4
elif s=="premium":
    bp*=1.2
if d>=30:
    bp*=0.9
elif d<7:
    bp*=1.25
if f:
    bp*=1.2
if a>60:
    bp*=0.85
print(bp)
```
